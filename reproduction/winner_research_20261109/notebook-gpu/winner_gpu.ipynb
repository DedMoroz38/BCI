Private research on all 10 people: 10 LOSO + later_last. Historical full baseline 62.7234% OOF is reused. No baseline retraining, parity checks, preliminary test suite or competition submission. Source and predictions are saved; public CV is exploratory.

```python
import os,sys,pathlib,io,base64,zipfile,subprocess,json
os.environ.update(OMP_NUM_THREADS='1',OPENBLAS_NUM_THREADS='1',MKL_NUM_THREADS='1',NUMBA_NUM_THREADS='1',MPLCONFIGDIR='/tmp/winner-mpl')
work=pathlib.Path('/tmp/winner-study');work.mkdir(parents=True,exist_ok=True)
with zipfile.ZipFile(io.BytesIO(base64.b64decode(__import__("pathlib").Path("local-bundle.base64").read_text()))) as z:z.extractall(work)
envdir=pathlib.Path('/tmp/winner-env')
subprocess.run([sys.executable,'-m','venv','--without-pip','--system-site-packages',str(envdir)],check=True)
python=str(envdir/'bin/python')
subprocess.run([sys.executable,'-m','pip','--python',python,'install','--quiet',*['numpy==2.2.6', 'scipy==1.15.3', 'scikit-learn==1.7.2', 'pandas==2.3.3', 'pyriemann==0.9', 'numba==0.61.2', 'geoopt==0.5.1']],check=True)
out=pathlib.Path('/kaggle/working/winner-gpu-results');out.mkdir(parents=True,exist_ok=True)
print((work/'EXECUTION.md').read_text(),flush=True)
print('Actual research starts next. No control fits, no pytest stage, no historical tolerance gate.',flush=True)
```

## Run the fixed research package

```python
with (out/'execution.log').open('w') as log:
    process=subprocess.Popen([python,'-u','gpu.py','--minutes','690'],cwd=work,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
    for line in process.stdout:
        print(line,end='',flush=True);log.write(line);log.flush()
    code=process.wait()
print('Process exit:',code,'; saved results:',out)
if code:raise RuntimeError('Runner stopped; completed predictions and failures are preserved in output')
```
