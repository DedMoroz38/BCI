# PROTEUS causal transduction: six research directions

118 fixed transduction settings, four stability/fusion variants and the saved fallback, each on Lag4 and Lag4+SWAD. 11 outer folds, 2706 evaluations; **no classifier retraining**. CPU execution preserves GPU quota. Target labels are used only for scoring. Public-CV results are exploratory. No reproduction gates or competition submission.

```python
import base64, io, os, pathlib, zipfile
for key in ('OMP_NUM_THREADS','OPENBLAS_NUM_THREADS','MKL_NUM_THREADS','NUMEXPR_NUM_THREADS'):
    os.environ[key]='1'
work=pathlib.Path('/tmp/proteus-transduction-study');work.mkdir(parents=True,exist_ok=True)
payload=base64.b64decode(__import__("pathlib").Path("local-bundle.base64").read_text())
with zipfile.ZipFile(io.BytesIO(payload)) as z:z.extractall(work)
out=pathlib.Path('/kaggle/working/transduction-results');out.mkdir(parents=True,exist_ok=True)
print('Ready: saved OOF priors, 6 research directions, CPU, 240-minute cap; no classifier training.')
print((work/'PROTOCOL.md').read_text())
```

## Run the full fixed grid

Completed folds are saved independently. No separate controls, pytest stage, numerical parity threshold or preliminary model training.

```python
import subprocess, sys
with (out/'execution.log').open('w') as log:
    process=subprocess.Popen([sys.executable,'-u','run.py','--workers','4','--minutes','240'],cwd=work,
        stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
    for line in process.stdout:
        print(line,end='',flush=True);log.write(line);log.flush()
    code=process.wait()
if code:raise RuntimeError(f'Study exited with code {code}; completed fold outputs remain in {out}')
```

## Results

OOF cell BA ranks complete ten-person results. Later-session scores and person deltas are reported separately. There are no epoch loss curves because no classifier is trained.

```python
from IPython.display import display, Image
report=out/'REPORT.md'
print(report.read_text() if report.exists() else (out/'completion.json').read_text() if (out/'completion.json').exists() else 'No complete folds; see execution.log and failures.json')
for name in ('top20.png','best_person_deltas.png'):
    if (out/name).exists():display(Image(filename=str(out/name)))
```
